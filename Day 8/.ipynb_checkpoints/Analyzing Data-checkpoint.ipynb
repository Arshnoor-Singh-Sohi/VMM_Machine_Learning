### Filtering Data from DataFrame

```python
# WAP to filter data of those students with marks greater than 90 in maths
```

```python
#Step 1 - Create a Filter list/series/array which contains only boolean values

filter1 = df['maths'] > 90
filter1
```

```python
#Step 2 - Apply filter list to the DataFrame

df[filter1]
```

```python
df[['name','roll','maths']]
```

```python
df[['name','roll','maths']][filter1]
```

```python
# WAP to filter students who have marks greater than 80 in physics and greater than 75 in chemistry
```

```python
filter2 = df['Physics'] > 80
filter3 = df['chemistry'] > 75
```

```python
df[filter2][filter3]
```

```python
df[filter2]
```

```python
df[filter3]
```

```python
df[df['biology']>60]                                                    
```

```python
df[df.name == 'Arshnoor']
```

### Query Function

```python
df.query("maths > 60")
```

```python
df.query("chemistry > 70 and Physics > 90")
```

```python
df['total'] = df['maths'] + df['Physics'] + df['chemistry'] + df['biology']
```

```python
df
```

```python
df['percentage(%)'] = df['total'] / 400 * 100
```

```python
df
```

```python
df[df['percentage(%)'] > 80]
```
