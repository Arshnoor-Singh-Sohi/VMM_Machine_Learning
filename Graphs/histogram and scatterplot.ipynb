```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
datapts = np.random.normal(170, 10, 250)
```

```python
plt.hist(datapts)
plt.show()
```

```python
plt.hist(datapts, bins=20)
plt.show()
```

```python
plt.hist(datapts, bins=30)
plt.show()
```

```python
plt.hist(datapts, bins=30, color='red')
plt.show()
```

```python
plt.hist(datapts, bins=30, color='red')
plt.title("Histogram Demo")
plt.show()
```

```python
plt.hist(datapts, bins=30, color='red')
plt.title("Histogram Demo")
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.show()
```

### Scatter Plot

```python
import random
```

```python
x = np.random.normal(170, 10, 20)
y = np.random.normal(50, 10, 20)
z = np.random.normal(250, 10, 20)
```

```python
plt.scatter(x, y, color='red')
plt.scatter(z, y, color='green')
plt.show()
```

```python
plt.scatter(x, y, color='red')
plt.scatter(z, y, c=np.random.randint(100, size=(20)), cmap='Dark2')
plt.show()
```
