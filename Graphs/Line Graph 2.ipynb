```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
years = np.array([2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022])

sales = np.array([1000000, 1500000, 1700000, 1800000, 2000000, 800000, 1200000, 1900000])
loss = np.array([200000, 500000, 300000, 500000,450000, 1000000, 800000, 300000])
```

```python
plt.plot(years, sales, marker='o')
plt.plot(years, loss, marker='o')
plt.legend(['sales', 'loss'])
plt.show()
```

```python
plt.plot(years, sales, marker='o')
plt.plot(years, loss, marker='o')
plt.legend(['sales', 'loss'], loc="upper right")
plt.show()
```

```python
plt.plot(years, sales, marker='o')
plt.plot(years, loss, marker='o')
plt.legend(['sales', 'loss'], loc="lower right")
plt.show()
```

```python
plt.plot(years, sales, marker='o')
plt.plot(years, loss, marker='o')
plt.legend(['sales', 'loss'], loc="lower right")
plt.grid()
plt.show()
```

```python
plt.plot(years, sales, marker='o')
plt.plot(years, loss, marker='o')
plt.legend(['sales', 'loss'], loc="lower right")
plt.grid(color='red')
plt.show()
```

```python
plt.plot(years, sales, marker='o')
plt.plot(years, loss, marker='o')
plt.legend(['sales', 'loss'], loc="lower right")
plt.grid(color='red', axis='x')
plt.show()
```

```python
plt.plot(years, sales, marker='o')
plt.plot(years, loss, marker='o')
plt.legend(['sales', 'loss'], loc="lower right")
plt.grid(color='red', axis='y')
plt.show()
```

### Subplotting

```python
plt.subplot(1,2,1) #Row Number, Total number of graphs in row, Graph Number
plt.plot(years, sales, marker='o')
plt.title("Sales Chart")

plt.subplot(1,2,2)
plt.plot(years, loss, marker='o')
plt.title("Loss Chart")

plt.show()
```
