```python
import matplotlib.pyplot as plt
import numpy as np
```

### Subplotting

```python
years = np.array([2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022])

sales = np.array([1000000, 1500000, 1700000, 1800000, 2000000, 800000, 1200000, 1900000])
loss = np.array([200000, 500000, 300000, 500000,450000, 1000000, 800000, 300000])
```

```python
plt.subplot(2,2,1) #Total number of rows, Total number of graphs in row, Graph Number
plt.plot(years, sales, marker='o')
plt.title("Sales Chart")

plt.subplot(2,2,2)
plt.plot(years, loss, marker='o')
plt.title("Loss Chart")

plt.show()
```

```python
plt.subplot(2,2,1) #Total number of rows, Total number of graphs in row, Graph Number
plt.plot(years, sales, marker='o')
# plt.title("Sales Chart")

plt.subplot(2,2,2)
plt.plot(years, loss, marker='o')
# plt.title("Loss Chart")

plt.subplot(2,2,3) #Total number of rows, Total number of graphs in row, Graph Number
plt.bar(years, loss)
# plt.title("Sales Chart")

plt.subplot(2,2,4)
plt.pie(loss, labels=years)
# plt.title("Loss Chart")

plt.show()
```
