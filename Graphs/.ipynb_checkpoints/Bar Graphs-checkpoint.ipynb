```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
years = np.array([2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022])

sales = np.array([1000000, 1500000, 1700000, 1800000, 2000000, 800000, 1200000, 1900000])
loss = np.array([200000, 500000, 300000, 500000,450000, 1000000, 800000, 300000])
```

```python
plt.bar(years, sales)
plt.show()
```

```python
plt.bar(years, sales)
plt.xlabel('Years')
plt.ylabel('Sales (in Millions)')
plt.title('Sales Chart')
plt.show()
```

```python
plt.bar(years, sales)
plt.xlabel('Years')
plt.ylabel('Sales (in Millions)')
plt.title('Sales Chart')
plt.grid()
plt.show()
```

```python
plt.bar(years, sales)
plt.xlabel('Years')
plt.ylabel('Sales (in Millions)')
plt.title('Sales Chart')
plt.grid(axis='y')
plt.show()
```

```python
plt.bar(years, sales, color='r')
plt.xlabel('Years')
plt.ylabel('Sales (in Millions)')
plt.title('Sales Chart')
plt.show()
```

```python
plt.barh(years, sales)
plt.xlabel('Years')
plt.ylabel('Sales (in Millions)')
plt.title('Sales Chart')
plt.show()
```

#### Height and Width parameters

```python
plt.bar(years, sales, width=0.1)
plt.xlabel('Years')
plt.ylabel('Sales (in Millions)')
plt.title('Sales Chart')
plt.show()
```

```python
plt.barh(years, sales, height=0.4)
plt.xlabel('Years')
plt.ylabel('Sales (in Millions)')
plt.title('Sales Chart')
plt.show()
```

### Pie Charts

```python
expenses = [2000, 2000, 3000, 500, 1000]
```

```python
plt.pie(expenses)
plt.show()
```

```python
mylabels = ['Fuels', 'F&B', 'Wardrobe', 'Donations', 'Fine']
plt.pie(expenses, labels=mylabels)
plt.show()
```

```python
mylabels = ['Fuels', 'F&B', 'Wardrobe', 'Donations', 'Fine']
myexplode = [0,0,0,0,0.3]
plt.pie(expenses, labels=mylabels, explode=myexplode)
plt.show()
```

```python
mylabels = ['Fuels', 'F&B', 'Wardrobe', 'Donations', 'Fine']
myexplode = [0,0,0,0,0.3]
plt.pie(expenses, labels=mylabels, explode=myexplode, shadow=True)
plt.show()
```

```python
mylabels = ['Fuels', 'F&B', 'Wardrobe', 'Donations', 'Fine']
myexplode = [0,0,0,0,0.3]
mycolors = ['green', 'blue', 'purple', 'cyan', 'red']
plt.pie(expenses, labels=mylabels, explode=myexplode, shadow=True, colors=mycolors)
plt.legend(title='My Expenses', loc='upper center')
plt.show()
```
