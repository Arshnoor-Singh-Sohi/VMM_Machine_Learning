```python
# Plot a graph of following columns from Pokemon.csv ,Attack, HitPoints, Total with different colors 
```

```python
df = pd.read_csv('Pokemon.csv')
df
```

```python
Grass = df['Type 1'] == 'Grass'
HP = df['HP']
Attack = df['Attack']
Total = df['Total']
plt.plot(HP)
plt.plot(Attack)
plt.plot(Total)
plt.ylabel('Grass Hit Point')
plt.show()
```

```python
for i in range(1,7):
    Generation = df.query(f"Generation=={i}")
    print(f"------------------Generation {i}-------------------")
    Pokemon_list = ['Water', 'Fire', 'Electric', 'Grass']
    for j in Pokemon_list:
        HitPoint = Generation['HP'][df['Type 1'] == j]
        Attack = Generation['Attack'][df['Type 1'] == j]
        Total = Generation['Total'][df['Type 1'] == j]
        plt.plot(HitPoint, label='Hit Point')
        plt.plot(Attack, label='Attack')
        plt.plot(Total, label='Total')
        plt.ylabel(j)
        plt.legend()
        plt.show()
```
