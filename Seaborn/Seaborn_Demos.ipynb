```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.get_dataset_names()
```

### Importing datasets

```python
crash_df = sns.load_dataset('car_crashes')
crash_df.head()
```

### Distribution Plot

```python
sns.distplot(crash_df['not_distracted'], kde=False, bins=15)

#kde -> Kernal Density Estimation
```

### Joint Plot

```python
sns.jointplot(x='speeding', y='alcohol', data=crash_df)
```

```python
sns.jointplot(x='speeding', y='alcohol', data=crash_df, kind='kde')
```

```python
sns.jointplot(x='speeding', y='alcohol', data=crash_df, kind='reg')
```

```python
sns.jointplot(x='speeding', y='alcohol', data=crash_df, kind='hex')
```

### KDE plot

```python
sns.kdeplot(x='alcohol', data=crash_df)
```

### Pair Plot

```python
sns.pairplot(crash_df)
```

### Importing Other Dataset

```python
tips_df = sns.load_dataset('tips')
tips_df
```

```python
sns.pairplot(tips_df)
```

```python
sns.pairplot(tips_df, hue='sex') # hue is used to categorise the Categorical Data
```

### Rug Plot

```python
sns.rugplot(tips_df['tip'])
```

#### Styling

```python
plt.figure(figsize=(16, 8))

sns.set_context('paper', font_scale=2)
# sns.set_context('talk', font_scale=2)
# sns.set_context('poster', font_scale=2)
sns.jointplot(x='speeding', y='alcohol', data=crash_df, kind='reg')
```

### Strip Plot

```python
sns.stripplot(x='day', y='total_bill', data=tips_df)
```

```python
plt.figure(figsize=(8,5))
sns.stripplot(x='day', y='total_bill', data=tips_df, palette='magma', hue='sex')
```

```python
plt.figure(figsize=(8,5))
sns.stripplot(x='day', y='total_bill', data=tips_df, palette='magma', hue='sex', dodge=True)
```

### Box Plot

```python
plt.figure(figsize=(14,9))

sns.boxplot(x='day', y='total_bill', data=tips_df, hue='sex')
```

### Voilin Plot

```python
# Violin Plot is a combination of Box Plot and KDE PLot
```

```python
plt.figure(figsize=(16,14))
sns.violinplot(x='day', y='total_bill', data=tips_df, hue='sex')
```

### Swarn Plot

```python
ax = plt.axes()
ax.set_facecolor("black")
sns.violinplot(x='day', y='total_bill', data=tips_df)
sns.swarmplot(x='day', y='total_bill', data=tips_df, color='white')
```

### Heatmaps

```python
plt.figure(figsize=(14, 10))

crash_matrix = crash_df.corr()

sns.heatmap(crash_matrix, cmap='Blues', annot=True)
```
