```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('stuData.csv')
df.iloc[9,2] = np.nan
df.iloc[9,3] = np.nan
df.iloc[0,4] = np.nan
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna()
```

### Dealing with Missing values

##### Method 1 -  Drop all rows with missing values

```python
df2 = df.dropna(axis=0)
df2
```

```python
df3 = df.dropna(axis=1)
df3
```

```python
df['name'].dropna()
```

#### Method 2 - Fill the Missing Values

```python
df11 = df.fillna("missing")
df11
```

```python
df['name'].fillna('missing', inplace=True)
df
```

```python
df13 = df.fillna(0)
df13
```

###### method and limit parameter of fillna()

```python
df
```

```python
df15 = df.fillna(method='ffill')
df15
```

```python
df16 = df.fillna(method='bfill')
df16
```

```python
df17 = df.fillna(0, limit=2)
df17
```

```python
df18 = df.fillna(method='ffill')
```

```python
df18
```

### Filling Missing Values with Mean, Median or Mode

```python
df['maths'].mean()
```

```python
df['physics'].median()
```

```python
df['chemistry'].mode()
```

```python
df['maths'].fillna(value=df['maths'].median(), inplace=True)
df
```

```python
df['physics'].fillna(value=df['physics'].median(), inplace=True)
df['chemistry'].fillna(value=df['chemistry'].median(), inplace=True)
df['biology'].fillna(value=df['biology'].median(), inplace=True)
df
```
