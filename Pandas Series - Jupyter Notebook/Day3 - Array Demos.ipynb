### Practice Program

Write a Program to generate array of 20 random numbers.

## Time and Space Comparison between list and array 

#### Space Comparison

```python
import sys

lst1 = list(range(1,1000001))

arr17 = numpy.arange(1, 1000001)

print('Size of List ', sys.getsizeof(lst1))
print('Size of Array ', sys.getsizeof(arr17))
```

#### Time Comparison

```python
import time

size = 10000000

list1 = list(range(size))
list2 = list(range(size))

array1 = numpy.array(list1)
array2 = numpy.array(list2)
```

```python
initial_list_time = time.time()
new_list = [(a*b) for a,b in zip(list1, list2)]
resultant_list_time = time.time()

print('Time taken by list in second ', resultant_list_time-initial_list_time)
```

```python
initial_array_time = time.time()

new_array = array1 * array2

resultant_array_time = time.time()

print('Time taken by array in second', resultant_array_time-initial_array_time)
```
