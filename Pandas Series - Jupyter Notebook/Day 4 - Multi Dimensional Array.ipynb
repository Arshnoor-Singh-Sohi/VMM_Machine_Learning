### Arithmatic Operations on Metrics

##### Addition

```python
met1 = np.array([[1,3,4], [2,4,3], [3,4,5]])
met2 = np.array([[1,3,4], [2,4,3],[1,2,4]])
```

```python
met3 = met1 + met2
```

```python
print(met3)
```

```python
met4 = np.add(met1, met2)
print(met4)
```

##### Subrtaction

```python
met1 = np.array([[2,4,6],[4,2,2],[2,4,2]])
met2 = np.array([[1,2,3],[2,1,1],[1,2,1]])
```

```python
met3 = met1 - met2

print(met3)
```

```python
met4 = np.subtract(met1, met2)
print(met4)
```

```python
met4.shape
```

##### Multiplication

```python
met1 = np.array([[1,2,3],[4,5,6]])
met2 = np.array([[10,11],[20,21],[30,31]])
```

```python
met3 = np.dot(met1, met2)
print(met3)
```
