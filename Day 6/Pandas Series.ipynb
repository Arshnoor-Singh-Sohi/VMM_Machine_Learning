```python
import numpy as np
import pandas as pd
```

### Creating Series from list

```python
lst = [12,13,14,15,16]
arr = np.array(lst)
s = pd.Series(lst)
```

```python
print(arr)
```

```python
print(s)
```

```python
s1 = pd.Series(['hello',1,2,3,4,5])
s1
```

```python
s2 = pd.Series([1.2,1,2,3,4,5,6])
s2
```

### Creating Series from Numpy Array

```python
lst1 = [1,2,3,2,4,2,4,6,3,7,5,54,34]
arr1 = np.array(lst1)
```

```python
arr1
```

```python
s3 = pd.Series(arr1)
s3
```

### Changing Series Index

```python
s4 = pd.Series(['hello','world','python'],index=[3,2,1])
s4
```

```python
s5 = pd.Series([1,'Rahil','BCA','78','B+'],index=['Rollno','Name','Class','Marks','Grade'])
s5
```

```python
# Length of index list must be same as length of the series.
```

### Creating Series from Dictionary

```python
d = {'Name':'Manav', 'Class':'Btech', 'Marks':50, 'Grade':'C'}

s7 = pd.Series(d)
s7
```

```python
d =[{'Name':'Manav', 'Class':'Btech', 'Marks':50, 'Grade':'C'},
    {'Name':'Manav', 'Class':'Btech', 'Marks':50, 'Grade':'C'}]
s8 = pd.Series(d)
s8
```

### Iterating over Series

```python
s9 = pd.Series([1,2,3,2,4,2,4,6,3,76,5,54,34])
s9
```

```python
for i in range(0, len(s9)):
    print(s9[i])
```

```python
for i in s9:
    print(i)
```
