```python
import pandas as pd
```

```python
df = pd.read_csv('Pokemon.csv')
df
```

```python
# WAP to filter pokemons with highest hitpoints of four categories water, fire, electric, grass in all generations
```

```python
for i in range(1,7):
    Generation = df.query(f"Generation=={i}")
    print(f"------------------Generation {i}-------------------")
    Pokemon_list = ['Water', 'Fire', 'Electric', 'Grass']
    for j in Pokemon_list:
        HitPoint = Generation['HP'][df['Type 1'] == j]
        max_hp = HitPoint.max()
        print(f"Maximum hit point of Generation {i} {j}  is " + str(max_hp))
```
