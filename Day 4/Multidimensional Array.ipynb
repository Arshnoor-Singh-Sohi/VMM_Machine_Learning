### Arithmatic Operations on Metrics

#### Addition

```python
met1 = np.array([[1,3,4], [2,4,3], [3,4,5]]) #matrix 1
met2 = np.array([[1,3,4], [2,4,3], [1,2,4]]) #matrix 2
```

```python
met3 = met1 + met2
print(met3)
```

```python
met4 = np.add(met1, met2)
print(met4)
```

#### Subtraction

```python
met1 = np.array([[2,4,6], [4,2,2], [2,4,2]])
met2 = np.array([[1,2,3], [2,1,1], [1,2,1]])
```

```python
met3 = met1 - met2
print(met3)
```

```python
met4 = np.subtract(met1, met2)
print(met4)
```

```python
#list is always 1-d but array can be 2-d or 3-d
```

#### Multiplication 

```python
met1 = np.array([[1,2,3], [4,5,6]])
met2 = np.array([[10,11], [20,21], [30,31]])
```

```python
met3 = met1*met2
```

```python
met3 = np.multiply(met1, met2)
```

```python
met3 = np.dot(met1, met2)
print(met3)
```

### Copy an Array

```python
arr10 = np.array([[10,11], [20,21], [30,31]])
```

```python
arr10
```

```python
arr10_copy = arr10
```

```python
arr10_copy #It is creating a reference not a copy
```

```python
arr10_copy[1][1] = 56
```

```python
arr10_copy
```

```python
arr10
```

```python
arr10_copyNew = arr10.copy() #Create a copy
```

```python
arr10_copyNew
```

```python
arr10_copyNew[0][0] = 345
arr10_copyNew
```

```python
arr10
```

```python
arr11 = arr10.view() #creates a reference
```

```python
arr11
```

```python
arr11[2][0] = 756
arr11
```
